# Minimal Presentations and Resolutions

Fix a parameter poset $P\subseteq\mathbb R^n$ with product order and a homological degree $d$. A multifiltered chain complex gives a persistence module

$$
M = H_d(C_*) = \ker(\partial_d)/\operatorname{im}(\partial_{d+1}).
$$

General algebraic background and the existence/uniqueness of minimal free
resolutions are surveyed by Botnan and Lesnick
{cite:p}`botnanIntroductionMultiparameterPersistence2023`. The 2-parameter
algorithms exposed by `multipers` follow the minimal-presentation framework and
`mpfree` speedups of Kerber and Rolle
{cite:p}`kerberFastMinimalPresentations2021`; the `2pac` backend uses the
cohomological clearing approach of Bauer, Lenzen, and Lesnick
{cite:p}`bauerEfficientTwoParameterPersistence2023`.

A free module generated at grade $a$ is $k_{\uparrow a}$, equivalently $S(-a)$ over $S=k[x_1,\ldots,x_n]$. A presentation is

$$
F_1 \xrightarrow{\phi_1} F_0 \to M \to 0,
\qquad M\cong \operatorname{coker}\phi_1.
$$

It is minimal when no same-grade identity summand remains in $\phi_1$. A free resolution continues left:

$$
\cdots \to F_2 \xrightarrow{\phi_2} F_1 \xrightarrow{\phi_1} F_0 \to M \to 0.
$$

For finitely presented $n$-parameter modules, minimal free resolutions exist, are unique up to isomorphism, and satisfy $F_i=0$ for $i\ge n+1$. Thus in two parameters only $F_0,F_1,F_2$ can be nonzero.

If

$$
F_i = \bigoplus_{a\in P} S(-a)^{\beta_{i,a}(M)},
$$

then $\beta_{i,a}(M)$ are the multigraded Betti numbers
{cite:p}`carlssonTheoryMultidimensionalPersistence2009`: generators,
relations, and higher syzygies.
