# Multicritical Free Implicit Resolution

A cell $\sigma$ in a multicritical bifiltration may have several incomparable minimal grades $G(\sigma)\subset P$. Its elementary module is

$$
U_G(p)=
\begin{cases}
k, & \exists g\in G:\ g\le p,\\
0, & \text{otherwise.}
\end{cases}
$$

The filtration is $k$-critical if $|G(\sigma)|\le k$ for every cell, and one-critical/free if every $|G(\sigma)|=1$.

The combinatorial presentation viewpoint of Chachólski,
Scolamiero, and Vaccarino
{cite:p}`chacholskiCombinatorialPresentationMultidimensional2017`
describes the homology of a multifiltration from the graded
boundary maps adjacent to a fixed degree. The free-resolution
algorithms used by `one_criticalify` are those of Bauer et al.
{cite:p}`bauerFastFreeResolutions2026`: they replace a
multicritical bifiltered chain complex by a quasi-isomorphic
one-critical/free chain complex.

For ordered grades $G=\{x_1,\ldots,x_k\}$, the path free resolution of $U_G$ is

$$
0\to \bigoplus_{i=1}^{k-1} k_{\uparrow(x_i\vee x_{i+1})}
\xrightarrow{p_1}
\bigoplus_{i=1}^k k_{\uparrow x_i}
\to U_G\to 0,
$$

with $p_1(r_i)=g_i+g_{i+1}$ over characteristic two. Replacing all multicritical cell modules by free resolutions gives a one-critical free chain complex $F_*$ and a quasi-isomorphism $F_*\to C_*$.

For degree $d$, the free implicit representation uses two consecutive boundary blocks

$$
A=\partial_{d+1}:F_{d+1}\to F_d,
\qquad
B=\partial_d:F_d\to F_{d-1},
\qquad BA=0,
$$

and represents

$$
H_d(p)=\ker B_p / \operatorname{im} A_p.
$$

In `multipers`, `mp.ops.one_criticalify` runs this multicritical-to-free bridge. `algo="path"` uses the path construction; `algo="tree"` uses the sparse shortcut/tree variant.
