# Signed Measure

For an integer-valued invariant $F:Q\to\mathbb Z$ on a finite poset $Q$, the
signed measure is the Mobius inversion $\mu_F$ characterized by

$$
F(q)=\sum_{u\le q}\mu_F(u).
$$

For [Hilbert](hilbert_function.ipynb) and Euler invariants, $Q=P$. For
[rank invariants](rank_invariant.ipynb), $Q=P\times P^{\mathrm{op}}$, where
the death coordinate uses the opposite order. `signed_measure` returns the atom
coordinates and integer weights of $\mu_F$; later vectorizations and distances
operate on this sparse signed measure. Rank signed measures can also be
projected to line barcodes as in the
[barcode-from-rank-signed-measure notebook](../example_zoo/barcode_from_rank_signed_measure.ipynb).
