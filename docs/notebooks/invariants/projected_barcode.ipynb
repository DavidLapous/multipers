# Projected Barcode

Let $X$ be an $n$-parameter filtration or filtered chain complex, and fix a
homological degree $d$. Its homology module $M=H_d(X)$ has a
gamma-linear projected barcode: the ordinary 1-parameter barcode obtained by
pushing the associated persistence sheaf forward along a positive linear form
{cite:p}`berkoukProjectedDistancesMultiparameter`
{cite:p}`fernandesComputationGammalinearProjected2025`.

In `multipers`, `projected_barcode` accepts either such a filtration/filtered
complex or an already computed full
[minimal free resolution](../ops/minimal_presentations_resolutions.ipynb)
$F_\bullet\to M$. If the input is a filtration, the wrapper first computes a
full free resolution of $H_d(X)$. If the input is already a resolution, it
reuses that algebraic model. In both cases, for a positive direction
$\gamma\in\mathbb R^n_{>0}$, each free generator at multigrade $a$ is assigned
scalar grade

$$
a^\gamma=\langle a,\gamma\rangle.
$$

Ordinary 1-parameter matrix reduction on this projected resolution gives the
projected barcode of $M$. This is not the ordinary barcode of the original
filtered chain complex after scalarizing its cell grades, and it is not the
restriction of $M$ to an affine line. For affine-line restrictions, see the
[fibered-barcode notebook](fibered_barcode.ipynb).
