# Rank Invariant

For a persistence module $M:P\to\mathrm{Vect}_k$, the rank invariant is

$$
\rho_M(a,b)=\operatorname{rank}\bigl(M(a\le b):M_a\to M_b\bigr),\qquad a\le b.
$$

The invariant lives on pairs $(a,b)$ with $a\le b$. Dense tensor output keeps
all pairs and fills incomparable entries with zero. For a 2-parameter grid of
shape $(m,n)$, `rank_invariant` returns shape $(m,n,m,n)$: two axes for the
source grade and two axes for the target grade. Its sparse decomposition is
covered in the [signed-measure notebook](signed_measure.ipynb), and line
slices are covered in the [fibered-barcode notebook](fibered_barcode.ipynb).
