# Hilbert Function

For a persistence module $M:P\to\mathrm{Vect}_k$, the Hilbert function is

$$
h_M(a)=\dim_k M_a.
$$

On a finite grid, its Hilbert signed measure $\mu_H$ is the Mobius inversion
characterized by the lower-set cumulative identity

$$
h_M(a)=\sum_{u\le a}\mu_H(u).
$$

`hilbert_function` computes this dense cumulative tensor on the requested grid.
Its sparse Möbius inversion is the Hilbert case of the
[signed-measure notebook](signed_measure.ipynb). If no grid is provided, the
function infers one from the filtration values.
