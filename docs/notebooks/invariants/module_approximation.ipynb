# Module Approximation

For a box $B=[\ell,u]\subset\mathbb R^n$, a module approximation is a finite
represented module $\widetilde M_B$ over $B$ used in place of the restricted
module $M|_B$.

With target error $\varepsilon$, the formal goal is an interleaving bound

$$
d_I(M|_B,\widetilde M_B)\le \varepsilon.
$$

`module_approximation` constructs $\widetilde M_B$ from line restrictions, as
in the [fibered-barcode notebook](fibered_barcode.ipynb). The box fixes the
coordinate range; `nlines` or `max_error` controls the sampling budget. The
output is a module object that can be plotted, sliced again, or converted into
vector representations. When starting from a chain complex, it is common to
first compute a [minimal presentation](../ops/minimal_presentations_resolutions.ipynb).
