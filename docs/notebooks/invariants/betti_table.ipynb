# Betti Table

For a [minimal free resolution](../ops/minimal_presentations_resolutions.ipynb)
of a persistence module,

$$
\cdots \to F_2 \to F_1 \to F_0 \to M \to 0,
\qquad
F_i=\bigoplus_{a\in P} k[P](-a)^{\beta_{i,a}},
$$

the multigraded Betti number $\beta_{i,a}$ is the multiplicity of the free
generator at grade $a$ in term $F_i$
{cite:p}`carlssonTheoryMultidimensionalPersistence2009`. In two parameters, a
minimal free resolution has no terms beyond $F_2$.

`betti_degrees` returns repeated grades $a$, one row per generator.
`betti_table` groups equal grades and returns pairs $(a,\beta_{i,a})$ term by
term, which is usually easier to read. Projecting those graded resolution terms
is covered in the [projected-barcode notebook](projected_barcode.ipynb).
